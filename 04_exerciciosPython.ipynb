## EXERCÍCIO 1

```python
# Exercício 1 - Utilizando o método input() crie um programa que receba a altura e o peso de uma pessoa e imprima na tela o IMC da pessoa 🏋 O programa deverá dizer se a pessoa está:

# Faixa do IMC	Classificação
# Abaixo de 18	Abaixo do peso
# Entre 18 e 25	Peso normal
# Entre 25 e 30	Sobrepeso
# Entre 30 e 35	Obesidade Grau I
# Entre 35 e 40	Obesidade Grau II
# 40 ou mais	Obesidade Grau III (Mórbida)

# IMC = Peso / altura x altura
```

```python
Altura = float(input("Digite sua altura em metros"))
Peso = float(input("Digite seu peso em kg"))

IMCaltura = (Altura * Altura) 

IMC = Peso / IMCaltura
print(IMC)

if IMC <= 18.5:
    print("Abaixo do peso!")

elif IMC >= 18.5 and IMC <= 24.9:
    print("Peso Normal!")

elif IMC > 25 and IMC <= 29.9:
   print("Sobrepeso!")

elif IMC >30 and IMC <= 34.9:
    print("Obesidade Grau I!")

elif IMC > 35 and IMC <= 39.9:
    print("Obesidade Grau II!")

elif IMC > 40:
    print("Obesidade Grau III!")
```

## EXERCICIO 5

```python
# Exercicio 5
#Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês. 
# Calcule e mostre o total do seu salário no referido mês, sabendo-se que são descontados 11% para o Imposto de Renda, 8% para o INSS e 5% para o sindicato, 
# faça um programa que nos dê:Salário bruto. Quanto pagou ao INSS. Quanto pagou ao sindicato. Quanto pagou de IR.
#O salário líquido.
```

```python
valor_h = float(input("Quanto você recebe por hora? "))

horas_mes = float(input("Quantas horas voce trabalhou no mês? "))

salario_bruto = valor_h * horas_mes

imposto_renda = salario_bruto * 0.11

inss = salario_bruto * 0.08

sindicato = salario_bruto * 0.05

salario_liq = salario_bruto - (inss + imposto_renda + sindicato)

print("Calculando os descontos para te dizer qual será seu salário liquido...")
print("\n")
print(f"Você pagará de Imposto de renda R$ {imposto_renda:.2f}")
print(f"Você pagará de inss R$ {inss:.2f}")
print(f"Sua contribuição ao sindicato será de R$ {sindicato:.2f}")
print(f"Sobrará para você miseros R$ {salario_liq:.2f}")
```

## EXERCICIO 6

```python
# Exercício 6
# Crie um programa que peça ao usuário 3 números, e apresente na tela esses números em ordem crescente. Exemplo: Usuário digitou 20, 5, 17
# Resposta = Os números digitados foram 5, 17 e 20.
```

```python
num1 = int(input("Digite um numero"))
num2 = int(input("Digite outro numero"))
num3 = int(input("Digite outro numero"))

if num1 <= num2 and num1 <= num3:
    print(num1)
    if num2 <= num3:
        print(num2)
        print(num3)
    else:
        print(num3)
        print(num2)

elif num2 <= num1 and num2 <= num3:
    print(num2)
    if num1 <= num3:
        print(num1)
        print(num3)
    else:
        print(num3)
        print(num1)

else:
    print(num3)
    if num1 <= num2:
        print(num1)
        print(num2)
    else:
        print(num2)
        print(num1)
```

```python
#Forma de resolver exercício 6 atavés da lista e do sort

num1 = int(input("Digite um numero"))
num2 = int(input("Digite outro numero"))
num3 = int(input("Digite outro numero"))

numeros = [num1, num2, num3]
numeros.sort()
print("Números em ordem crescente:", numeros)
```

## EXERCÍCIO 7

```python
# Exercícios 7
# Peça para o usuário digitar um ano, e informe se ele é um ano bissexto ou não.

#ANO BISSEXTO - PARA ELE SER TEM QUE SER:
# - DIVISIVEL POR 4
# - NÃO DIVISÍVEL POR 100
# - A NÃO SER QUE ELE SEJA DIVISIVEL POR 400
```

```python
ano = int(input("Digite um ano para saber se ele é bissexto: "))

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(f"O ano {ano} é bissexto!")
else:
    print(f"O ano {ano} não é bissexto")
```
