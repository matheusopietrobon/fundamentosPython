```python
# FUNCAO COM ARGUMENTO

def greetin(name):
    print(f"Hello {name}")

nome = input("Digite seu nome:")

greetin(nome)
```

```python
# FUNCAO IMPRIMA O MAIOR NUMERO

def maior (x,y):
    if x > y:
        print(x)
    else:
        print(y)

maior(10, 15)
```

```python
# FUNCAO USANDO RETURN

def addNum(num1, num2):
    return num1 + num2


resultado = addNum(10,10)
resultado2 = addNum(20,20)

print("Printando o que a função devolveu", resultado)
print("Printando o que a função devolveu", resultado2)
```
