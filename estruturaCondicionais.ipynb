```python
temperatura = float(input("Digite a temperatura em gaus celsius"))
if temperatura >= 30:
    print("Está quente!")

elif temperatura <= 15:
    print("Está frio!")

else:
    print("Clima está agradável!")



# print("Esse print acontece independente do if")
```
