```python
minhaString = "hello  world!"
outraString = " world!"

# Acessar Caracteres Individualmente:
minhaString[0] #H
minhaString[1] #e
minhaString[2] #l
minhaString[3] #l
minhaString[4] #o
print(minhaString[0])

# Concatenar Strings:
novaString = minhaString + outraString
print(novaString)

# Repetir Strings:
repetida = minhaString * 5
print(repetida)

# Fatiar Strings:
minhaString[0:5] # "Hello"
minhaString[2:] # "llo"
minhaString[: :2] # "Hlo" (foi de 2 em 2 casas)
minhaString[-5] # "H" (de trás pra frente não conta o zero)
minhaString[-1] # "o" (ultimo caractere)
minhaString[: :-1] # "olleH"

```

```python

print(type(minhaString))
print(dir(minhaString))
```

```python
len(minhaString)
len("Hello World!")
minhaString.upper() # Tudo maiusculo
minhaString.lower() # Tudo minusculo
minhaString.capitalize() # Coloca apenas a primeira letra da primeira palavra em maiusculo
novaString.title() # Coloca as primeiras letras das palavras em maiusculo
minhaString.strip() # Corta os espaços em branco antes da primeira palavra e depois da ultima palavra.
minhaString.split() # Divide a frase em palavras como uma lista
minhaString.replace("l","x").replace("o","x").replace("h","x").replace("e","x").replace("w","x").replace("r","x").replace("d","x").replace("!","x")
minhaString.find("o")
minhaString.count("l") # Conta quantas letras tem na frase
# print(minhaString.count("l")+ minhaString.count("o"))
minhaString.startswith("h") # Primeira letra ou palavra qu inicia (vedadeiro ou falso)
minhaString.endswith("!") # Ultima letra ou palavra qu inicia (vedadeiro ou falso)


print("ell" in minhaString) # India se tem ou nao o que vc procura (verdadeiro ou falso)



```
