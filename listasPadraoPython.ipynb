```python
# ==============================================================================
# ATIVIDADE PRÁTICA: TRABALHANDO COM LISTAS EM PYTHON
# ==============================================================================
# As listas são estruturas de dados mutáveis e ordenadas que permitem armazenar
# múltiplos elementos em uma única variável.
# ==============================================================================

# ------------------------------------------------------------------------------
# EXEMPLO 1: Criando e Acessando Listas
# ------------------------------------------------------------------------------
print("=== EXEMPLO 1: Criando e Acessando Listas ===")
frutas = ["maçã", "banana", "laranja", "uva", "manga"]

# Acessando por índice (inicia em 0)
print(f"Primeira fruta: {frutas[0]}")
print(f"Última fruta: {frutas[-1]}")

# Fatiamento (Slicing)
print(f"Duas primeiras frutas: {frutas[:2]}")
print()


# ------------------------------------------------------------------------------
# EXEMPLO 2: Adicionando e Removendo Elementos
# ------------------------------------------------------------------------------
print("=== EXEMPLO 2: Adicionando e Removendo Elementos ===")
compras = ["arroz", "feijão"]

# Adicionando elementos
compras.append("ovo")         # Adiciona ao final
compras.insert(1, "macarrão") # Insere na posição/índice 1
print(f"Lista de compras atualizada: {compras}")

# Removendo elementos
compras.remove("feijão")      # Remove pelo valor
item_removido = compras.pop() # Remove e retorna o último elemento
print(f"Item removido com pop(): {item_removido}")
print(f"Lista final de compras: {compras}")
print()


# ------------------------------------------------------------------------------
# EXEMPLO 3: Funções Úteis com Listas Numéricas
# ------------------------------------------------------------------------------
print("=== EXEMPLO 3: Operações com Números ===")
notas = [7.5, 8.0, 6.5, 9.0, 10.0, 5.5]

print(f"Notas dos alunos: {notas}")
print(f"Quantidade de notas: {len(notas)}")
print(f"Maior nota: {max(notas)}")
print(f"Menor nota: {min(notas)}")
print(f"Soma das notas: {sum(notas)}")

# Calculando a média
media = sum(notas) / len(notas)
print(f"Média da turma: {media:.2f}")

# Ordenando a lista
notas.sort() # Ordena do menor para o maior (crescente)
print(f"Notas ordenadas: {notas}")
print()


# ------------------------------------------------------------------------------
# EXEMPLO 4: Percorrendo Listas com Loop (for)
# ------------------------------------------------------------------------------
print("=== EXEMPLO 4: Iterando sobre uma Lista ===")
alunos = ["Ana", "Bruno", "Carla", "Daniel"]

for i, aluno in enumerate(alunos, start=1):
    print(f"Aluno {i}: {aluno}")
print()


# ==============================================================================
# DESAFIO PRÁTICO PARA VOCÊ RESOLVER
# ==============================================================================
#EXERCÍCIO:
#1. Crie uma lista chamada 'numeros' contendo os valores: [12, 45, 7, 23, 89, 34].
#2. Adicione o número 50 ao final da lista.
#3. Calcule e exiba a soma de todos os números pares da lista.
#4. Crie uma nova lista apenas com os números maiores que 20.

#(Descomente o código abaixo para testar sua solução!)

# SEU CÓDIGO AQUI:
# numeros = [12, 45, 7, 23, 89, 34]
# numeros.append(50)
# pares = [n for n in numeros if n % 2 == 0]
# maiores_que_20 = [n for n in numeros if n > 20]
# print("Soma dos pares:", sum(pares))
# print("Maiores que 20:", maiores_que_20)
```
