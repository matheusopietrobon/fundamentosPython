```python
list_x = ["e", "c", "a", "d", "b"]
list_y = ["1", "5", "3", "2", "4"]
lista_ordenada = sorted(list_x)
print(lista_ordenada)

list_x.sort()
list_y.sort()
print(list_x)
print(list_y)

list_y.reverse ()
print(list_y)

list_y.copy()
print(list_y)

import random
random.shuffle(list_x)
print(list_x)

list_z = list_x + list_y
random.shuffle(list_z)
print(list_z)

list_z.insert(0, "OIE")
print(list_z)

list_z.insert(4, "OIE")
print(list_z)

list_z.insert(8, "OIE")
print(list_z)

list_z.insert(13, "OIE")
print(list_z)
```
