### Operadores Matemáticos em Python

```python
# soma em python
print("Esse é o primeiro print (soma)", 1+5)
# subtracao em python
print("Esse é o segundo print (subtracao)", 4-3)
# muliplicação em python
print("Esse é o terceiro print (multiplicacao)", 10*5)
# divisão em python
print("Esse é o quarto print (divisão)", 100/5)
# divisão inteira
print("Esse é o quinto print (divisão inteira)", 10//3)
# resto da divisão
print("Esse é o sexto print (%)", 15%2)
# potência
print("Esse é o sétimo print (potência ao quadrado)", 5**3)
# raiz quadrada
print("Esse é o oitavo print (raiz quadrada)", 81**(1/2))
print("Você conseguiu com sucesso...")
print("PARABÉNS!!")

```

### Operadores Boolanos

```python
# operador de igual
print("Está correta minha resposta? 5 é igual a 3?", 5 == 3)
# obs: um "=" é atribuição / dois "==" é comparação

# operador diferente
print("Está correta minha resposta? 5 é diferente de 4?",5 != 4)

# maior/igual que
print("5 é maior/igual que 5?", 5>=5)
print("5 é maior que 5?", 5>5)

# menor/igual que
print("6 é menor/igual que 6?", 6<=6)
print("6 é menor que 6?", 6<6)
```
