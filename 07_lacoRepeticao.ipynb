```python
# WHILE

count = 0
while count <= 5:
    print(f"o valor do contador é {count}")
    count += 1

print("Aqui o while acabou!")
```

```python
# USO DO CONTINUE

count = 0
while count <= 10:
    print(f"o valor do contador é {count}")
    count += 1
    continue
    print("teste do cotinue")

print("Aqui o while acabou!")
```

```python
# USO DO BREAK

count = 0
while True:
    if count <= 10:
        print(f"O valor do contador é {count}")
        count += 1
    else:
        print("Aqui executa antes do while")
        break

print("Carregando....")
#time.sleep(3)
print("o fora do while, break me quebrou!!!")
```

```python
# LAÇO DE REPETICAO "FOR"

for i in range(-3, 3):
    print(i)

print("DIVISÃO DE CÓDIGOS")

for i in range(-5, 5, 2):
    print(i)
```
